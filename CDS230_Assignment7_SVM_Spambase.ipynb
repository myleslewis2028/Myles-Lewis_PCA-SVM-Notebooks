Background information : This dataset represents "Emails", in specific, this shows the classification task for this dataset whether the given email is spam or not. This dataset contains total 57 features and 1 target column. Therefore, in this colab, we will discover and evaluate SVM model to find an optimal decision boundary with it's ability in accuracy, precision, and recall.

Data source link : https://archive.ics.uci.edu/dataset/94/spambase

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn import svm
from sklearn import metrics
```

```python
spambase_df = pd.read_csv('spambase.data', header=None)
spambase_df
```

```python
spambase_df.shape
```

By using the spambase_df.shape, it shows the 4601 rows and 58 columns in total.

```python
X = spambase_df.iloc[:, :-1]
y = spambase_df.iloc[:, -1]
```

```python
print("Features shape: ", X.shape)
print("Labels distribution:\n", y.value_counts())
```

This shows that the code of "spambase_df.shape" shows the whole columns with 58 columns, but the code with "X.shape" shows the Features, X which exclude the target column and as a result of it, it shows 57 columns.

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=100)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

During the vibe coding, it says that standardscaler is necessary for improving the accurate ability of SVM. So, for getting better and organized result from the SVM mode, adjusting the scaler as standardscaler is necessaray.

```python
clf = svm.SVC(kernel='rbf', random_state=100)
clf.fit(x_train_scaled, y_train)
```

Creating SVM model for SVM test and prediction

```python
y_pred = clf.predict(x_test_scaled)
```

As we've created the SVM model for this dataset, we could predict the result of test dataset, in specific "Accuracy, Precision, and Recall"

```python
print("Accuracy: ", metrics.accuracy_score(y_test, y_pred))
print("Precision: ", metrics.precision_score(y_test, y_pred))
print("Recall: ", metrics.recall_score(y_test, y_pred))
```

This code shows the accuracy, precision, recall which demonstrate how good the SVM model is, or the ability from it.

In conclusion, from the prediction it shows the result of test data with


Evaluation Results :

0.932 Accuracy, which shows approximately 93.3% of the total test emails were correctly classified.
0.936 Precision, which shows that out of all emails predicted as spam, 93.6% were actually spam, ensuring a low false-positive rate.
0.898 Recall, which shows the model successfully identified about 89.8% of all actual spam emails in the test set.

In addition, the step of StandardScaler prior to training played a crucial role in preventing features with larger scales from dominating the distance-based margin calculations, leading to a robust and highly accurate classification model.
